## *Opdracht 1*: Sanity check 3

```python
# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
epsilon0 = 8.854187817e-12 # F/m

def vlakkeplaat(A, d, epsilon1 =1):
    # A: oppervlakte van de platen (m²)
    # d: afstand tussen de platen (m)
    # epsilon0: permittiviteit van vacuum (F/m)
    # epsilon1: relatieve permittiviteit van het materiaal tussen de platen (dimensionless)
    # bij vacuum en lucht geld epsilon1 = 1
    global epsilon0
    C = (epsilon0 * epsilon1* A)/ d
    return C
```

```python
# Gebruik je formule om uit te rekenen wat je verwacht dat de capaciteit van jullie vlakke plaat condensatoren
# gaat zijn.

C = [vlakkeplaat(2.0, 0.001), vlakkeplaat(4.7, 0.001),vlakkeplaat(1.2, 0.001)]


print(f"Berekende capaciteit plaatcondensator groep 52 is {C[0]:.3e} F")
print(f"Berekende capaciteit plaatcondensator groep 52 is {C[1]:.3e} F")
print(f"Berekende capaciteit plaatcondensator groep 52 is {C[2]:.3e} F")
```
