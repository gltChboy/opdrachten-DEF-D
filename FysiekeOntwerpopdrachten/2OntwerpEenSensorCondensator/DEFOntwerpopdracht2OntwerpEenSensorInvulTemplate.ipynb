## *Opdracht 1*: Sanity check 3

```python
import numpy as np
import matplotlib.pyplot as plt

# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!
epsilon0 = 8.854187817e-12 # F/m

def vlakkeplaat(A, d, epsilon1 =1):
    # A: oppervlakte van de platen (m²)
    # d: afstand tussen de platen (m)
    # epsilon0: permittiviteit van vacuum (F/m)
    # epsilon1: relatieve permittiviteit van het materiaal tussen de platen (dimensionless)
    # bij vacuum en lucht geld epsilon1 = 1
    global epsilon0
    C = (epsilon0 * epsilon1* A)/ d
    return C
```

```python
# Gebruik je formule om uit te rekenen wat je verwacht dat de capaciteit van jullie vlakke plaat condensatoren
# gaat zijn.

C = [vlakkeplaat(2.0, 0.001), vlakkeplaat(4.7, 0.001),vlakkeplaat(1.2, 0.001)]

print(f"Berekende capaciteit plaatcondensator groep 52 is {C[0]:.3e} F")
print(f"Berekende capaciteit plaatcondensator groep 52 is {C[1]:.3e} F")
print(f"Berekende capaciteit plaatcondensator groep 52 is {C[2]:.3e} F")
```

```python
#berekende
epsilon1 = 2 #permabiliteit van het tussenmatereaal (volgens wikipedia)
A = 0.1**2  #m^2
d = 0.00011    #m

C = vlakkeplaat(A, d, epsilon1)
print(f"Berekende capaciteit plaatcondensator groep 52 is {C:.3e} F")

R_shunt = 1000 #Ohm
f = 100000 #Hz
V_out = C * R_shunt * 2 * np.pi * f
print(f"De output spanning van de sensor is {V_out:.3e} V")

#gemete onderdeel
V_out_gemeten = 4.46#V
print(f"De gemeten output spanning van de sensor is {V_out_gemeten:.3e} V")
C = V_out_gemeten/(R_shunt * 2 * np.pi * f)
print(f"De gemeten capaciteit van de sensor is {C:.3e} F")
```

## Opdracht 7: Plot de kalibratiegrafiek.

```python
from scipy.optimize import curve_fit

m = np.array([0, 0.5145, 0.5981, 0.7202, 0.9348, 1.1196, 1.4345, 1.8805, 2.3896, 3.213]) #kg
F = np.array([51.1, 74.0, 82, 85, 218, 298, 488, 504, 580, 650]) * 1e-12 

def model(m ,a,b,c):
    return a/(b + m * c)

popt, pcov = curve_fit(model, m, F)
print(popt)
#plot
plt.figure(figsize=(8,5))
plt.xlabel("Massa op de sensor (kg)")
plt.ylabel("Capaciteit van de sensor (pF)")
plt.plot(m, F * 1e12, ".", label="gemeten data")
plt.plot(m, model(m, *popt) * 1e12, label="fit")
plt.grid()
plt.errorbar(m, F * 1e12, yerr=10*m, fmt="none")
plt.show()


```
