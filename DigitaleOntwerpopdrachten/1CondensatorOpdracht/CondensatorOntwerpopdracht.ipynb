````python
#3 weet niet 100% zeker of dit goed gaat dus even bij TA nachecken
#maar het komt wel heel netjes uit met exact "b = -1"

def fit(r, V):
    popt, pcov = curve_fit(model, r, V)  # pas de functie 'model' aan op de data van Ez voor de index die het dichtst bij z=0 ligt
    a = popt[0]  # de waarde van a in de model functie
    b = popt[1]  # de waarde van b in de model functie
    c = popt[2]  # de waarde van c in de model functie
    return a, b, c, pcov

def model(r, a, b, c):
    #2 functie vanwege een knik in de data
    # eerste stuk is linear, tweede stuk met 1/r
    global knik
    return np.where(r <= r[knik], 
                    c * r, 
                    a * r**b)

#index voor de max waarde van V
#is nodig om te bepalen waar de knik in de data zit, 
#zodat we tot waar de data linear is en vanaf waar 1/r gedrag "vertoont"
knik = np.argmax(V[idx_z0])
print(f"maximale punt van V: {knik}")

a, b, c, pcov = fit(r, V[idx_z0])
print(a, b, c)

r_fit = np.linspace(min(r), max(r), 100)
V_fit = model(r_fit, a, b, c)

plt.figure(figsize=(10, 10))
plt.loglog(r, V[idx_z0], label='Data')
plt.loglog(r_fit, V_fit, "--",label='Fit: a/r + b')
plt.xlabel("r (m)")
plt.ylabel("V (V)")

plt.grid()
plt.show()
````

#### Vraag 5a: Functies voor de capaciteit van enkele opdrachten

1. Vind de formule voor de capaciteit van een vlakke plaat condensator en een cilinder condensator, onder de aanname dat de afstand tussen de platen veel kleiner is dan het oppervlakte van de condensator (vlakke plaat aanname).
2. Schrijf een python functie die de relevante parameters als input neemt en de capaciteit van een vlakke plaat condensator als output heeft.
3. Vind de formule voor de capaciteit van een cilinder condensator onder de aanname dat de afstand tussen de binnen en buiten wand klein is.
4. Schrijf een python functie die de relevante parameters als input neemt en de capaciteit van een cilinder condensator als output heeft.

````python
#epsilon1 : Relatieve permittiviteit van het materiaal (vacuüm , lucht = 1)

def vlakkeplaat(epsilon0, A, d, epsilon1 =1):
    # A: oppervlakte van de platen (m²)
    # d: afstand tussen de platen (m)
    # epsilon0: permittiviteit van vacuum (F/m)
    C = (epsilon0 * epsilon1* A)/ d
    return C

def cilinder(epsilon0, a, b, L, epsilon1= 1):
    # a: straal binnenste cilinder
    # b: straal buitenste cilinder
    # L: lengte van de cilinder
    # epsilon0: permittiviteit van vacuum (F/m)
    C = (2 * np.pi * epsilon0 * epsilon1 * L) / (np.log(b/a))
    return  C
````
